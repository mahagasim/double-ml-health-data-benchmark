# 01 — Theory, causal target, and simulated DGP

This notebook is the conceptual entry point. The primary target is the constant structural effect $\theta_0=1$ in $Y=D\theta_0+g_0(X)+\varepsilon$. Because the effect is constant, it is also the ATE.

The DGP encodes consistency, conditional exchangeability, positivity (stressed in Scenario D), and no interference. DML addresses estimation with complex nuisance functions; it does not repair identification failures.

```python
import numpy as np
import pandas as pd
from dml_health_benchmark.dgp import SCENARIOS, generate_data
```

```python
rows=[]
for name, config in SCENARIOS.items():
    data=generate_data(config, seed=20260811)
    rows.append({'scenario':name,'N':config.n,'p':config.p,'true_ATE':data.true_ate,'treatment_prevalence':data.d.mean(),'share_e_outside_005_095':np.mean((data.e_true<.05)|(data.e_true>.95))})
pd.DataFrame(rows)
```

## Scenario progression

A→B increases dimensionality while retaining sparse linear nuisance structure. B→C adds nonlinearities and interactions. C→D holds the nonlinear outcome structure fixed while worsening overlap.

The primary rich nuisance dictionary applies raw terms, centered squares, sine transforms, centered threshold indicators, and adjacent interactions to every covariate. With $p=400$, this yields $p^*=1999>N=1000$.
